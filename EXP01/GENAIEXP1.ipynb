```python
from transformers import pipeline
```

```python
generator = pipeline("text-generation", model="gpt2")
print("GPT-2 text generation pipeline loaded successfully.")
```

```python
prompt = "Artificial Intelligence will shape the future of"

# Words/content to avoid
bad_words = ["violence", "war"]
bad_words_ids = [
    generator.tokenizer.encode(word, add_special_tokens=False)
    for word in bad_words
]

# Token IDs for beginning and ending constraints
bos_token_id = generator.tokenizer.encode(
    " Technology", add_special_tokens=False
)[0]

eos_token_id = generator.tokenizer.eos_token_id

result = generator(
    prompt,
    max_length=80,
    min_length=30,
    temperature=0.8,
    do_sample=True,
    top_k=50,
    top_p=0.95,
    repetition_penalty=1.2,
    no_repeat_ngram_size=2,
    bad_words_ids=bad_words_ids,
    forced_bos_token_id=bos_token_id,
    forced_eos_token_id=eos_token_id,
    num_return_sequences=1
)

print("Prompt:")
print(prompt)

print("\nGenerated Text:")
print(result[0]["generated_text"])

print("\nParameters Used:")
print("bad_words_ids       : Avoids specified words/content.")
print("forced_bos_token_id : Forces a specific beginning token.")
print("forced_eos_token_id : Forces an end-of-sequence token.")
print("no_repeat_ngram_size: Prevents repeated word sequences.")
```

```python
prompts = [
    "Artificial Intelligence will shape the future of",
    "Generative AI is transforming education because",
    "Sustainability can be improved through",
    "The future of healthcare will depend on",
    "Technology and human creativity together can"
]

for i, prompt in enumerate(prompts, 1):
    result = generator(
        prompt,
        max_length=60,
        temperature=0.8,
        do_sample=True,
        top_k=50
    )

    print(f"\n--- Prompt {i} ---")
    print("Input:", prompt)
    print("Output:", result[0]["generated_text"])
```

```python
prompt = "Artificial Intelligence will shape the future of"

settings = [
    {"max_length": 50, "temperature": 0.5, "top_k": 30},
    {"max_length": 60, "temperature": 0.8, "top_k": 50},
    {"max_length": 80, "temperature": 1.0, "top_k": 70}
]

for i, setting in enumerate(settings, 1):
    result = generator(
        prompt,
        max_length=setting["max_length"],
        temperature=setting["temperature"],
        do_sample=True,
        top_k=setting["top_k"]
    )

    print(f"\n--- Experiment {i} ---")
    print("Parameters:", setting)
    print("Generated Text:")
    print(result[0]["generated_text"])
```

## Step 9: Observations
Compare the generated responses.

- Lower temperature generally produces more predictable text.
- Higher temperature generally produces more varied and creative text.
- Increasing `max_length` allows longer responses.
- `top_k` controls the number of likely next-token candidates considered during generation.
- Different prompts produce different generated responses.
- The generated text may change each time because sampling is enabled with `do_sample=True`.
